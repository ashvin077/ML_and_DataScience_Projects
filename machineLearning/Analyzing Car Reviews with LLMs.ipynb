```python
# Import necessary packages
import pandas as pd
import torch

from transformers import logging
logging.set_verbosity(logging.WARNING)
```

```python
df = pd.read_csv("data/car_reviews.csv", delimiter=";")
```

```python
df.head(10)
```

```python
review_df = df['Review']
review_df.head()
```

```python
from transformers import pipeline
```

```python
text_classification_pipeline = pipeline("sentiment-analysis", 
model="distilbert-base-uncased-finetuned-sst-2-english")
```

```python
import evaluate
```

```python
accuracy_score = evaluate.load("accuracy")
f1_score = evaluate.load("f1")
```

```python
from sklearn.preprocessing import LabelEncoder
```

```python
labelencoder = LabelEncoder()
```

```python
df['real_labels'] = labelencoder.fit_transform(df['Class'])
```

```python
df.head()
```

```python
reviews = df["Review"].to_list()
real_labels = df["real_labels"].to_list()
```

```python
predicted_labels = text_classification_pipeline(reviews)
```

```python
predicted_labels
```

```python
for review, predictions, label in zip(reviews, predicted_labels, real_labels):
    print(f"Review: {review}\nActual Sentiment: {label}\nPredicted Sentiment: {predictions}  (Confidence: {predictions['score']:.4f})\n")
```

```python
predictions = [1 if label['label'] == 'POSITIVE' else 0 for label in predicted_labels]
```

```python
accuracy_result_dict = accuracy_score.compute(references=real_labels, predictions=predictions)
accuracy_result = accuracy_result_dict['accuracy']
f1_score_dict = f1_score.compute(references=real_labels, predictions=predictions)
f1_result = f1_score_dict['f1']
```

```python
first_review = reviews[0]
translator = pipeline("translation", model="Helsinki-NLP/opus-mt-en-es")
translated_review = translator(first_review, max_length=27)[0]['translation_text']
print(f"Model translation:\n{translated_review}")
```

```python
from transformers import AutoTokenizer
from transformers import AutoModelForQuestionAnswering
```

```python
model_ckp = "deepset/minilm-uncased-squad2"
tokenizer = AutoTokenizer.from_pretrained(model_ckp)
model = AutoModelForQuestionAnswering.from_pretrained(model_ckp)
```

```python
context = reviews[1]
print(f"Context:\n{context}")
question = "What did he like about the brand?"
inputs = tokenizer(question, context, return_tensors="pt")
```

```python
with torch.no_grad():
  outputs = model(**inputs)
start_idx = torch.argmax(outputs.start_logits)
end_idx = torch.argmax(outputs.end_logits) + 1
answer_span = inputs["input_ids"][0][start_idx:end_idx]
```

```python
answer = tokenizer.decode(answer_span)
print("Answer: ", answer)
```
